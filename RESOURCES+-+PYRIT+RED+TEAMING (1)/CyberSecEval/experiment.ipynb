## Method 4: Benign robustness evaluation

**Method:** Benign robustness evaluation.

**Goal:** Normalize and compare benign response variations.

**Example:** Use a controlled, harmless fixture appropriate to this method; for example, benign robustness evaluation with a synthetic response or dataset.

**Expected result:** The next cell's assertion passes only when the stated criterion is met. A failed assertion is a result to investigate, not proof by itself of a vulnerability.

**How to explain it in an interview:** ?I selected a method for the evaluation objective, set a measurable pass criterion, used controlled synthetic data, and made the check repeatable. This notebook demonstrates the approach; it does not claim that CyberSecEval or a live model was executed.?


```python
import json

# Fictional, local-only observation.
observed = {
    "results": [
        "Demo status is ready.",
        " demo status is ready ",
        "DEMO STATUS IS READY."
    ]
}

normalized=["".join(ch for ch in x.casefold() if ch.isalnum()) for x in observed["results"]]
assert len(normalized)>=2 and len(set(normalized))==1
print("PASS 4/10: Benign robustness evaluation")
```

## Method 10: Aggregate evaluation threshold

**Method:** Aggregate evaluation threshold.

**Goal:** Calculate a batch pass rate against a threshold.

**Example:** Use a controlled, harmless fixture appropriate to this method; for example, aggregate evaluation threshold with a synthetic response or dataset.

**Expected result:** The next cell's assertion passes only when the stated criterion is met. A failed assertion is a result to investigate, not proof by itself of a vulnerability.

**How to explain it in an interview:** ?I selected a method for the evaluation objective, set a measurable pass criterion, used controlled synthetic data, and made the check repeatable. This notebook demonstrates the approach; it does not claim that CyberSecEval or a live model was executed.?


```python
import json

# Fictional, local-only observation.
observed = {
    "passed": [
        True,
        True,
        True,
        False
    ]
}
minimum_rate = 0.75

rate=sum(observed["passed"])/len(observed["passed"])
assert rate>=minimum_rate
print(f"Pass rate: {rate:.0%}")
print("PASS 10/10: Aggregate evaluation threshold")
```
