```python
import json

# Fictional, local-only observation.
observed = {
    "results": [
        "Birds have feathers.",
        "Rivers carry water.",
        "Winter is a season."
    ]
}
expected_terms = ["Birds", "water", "season"]

assert len(observed["results"])==len(expected_terms)
assert all(term.casefold() in text.casefold() for text,term in zip(observed["results"],expected_terms))
print("PASS 1/10: Task dataset sample")
```

## Method 10: Evaluation pass-rate metric

**Method:** Evaluation pass-rate metric.

**Goal:** Calculate a batch pass rate against a threshold.

**Example:** Use a controlled, harmless fixture appropriate to this method; for example, evaluation pass-rate metric with a synthetic response or dataset.

**Expected result:** The next cell's assertion passes only when the stated criterion is met. A failed assertion is a result to investigate, not proof by itself of a vulnerability.

**How to explain it in an interview:** ?I selected a method for the evaluation objective, set a measurable pass criterion, used controlled synthetic data, and made the check repeatable. This notebook demonstrates the approach; it does not claim that Inspect AI or a live model was executed.?


```python
import json

# Fictional, local-only observation.
observed = {
    "passed": [
        True,
        True,
        True,
        False
    ]
}
minimum_rate = 0.75

rate=sum(observed["passed"])/len(observed["passed"])
assert rate>=minimum_rate
print(f"Pass rate: {rate:.0%}")
print("PASS 10/10: Evaluation pass-rate metric")
```
