## Method 10: Scenario-set consistency

**Method:** Scenario-set consistency.

**Goal:** Normalize and compare benign response variations.

**Example:** Use a controlled, harmless fixture appropriate to this method; for example, scenario-set consistency with a synthetic response or dataset.

**Expected result:** The next cell's assertion passes only when the stated criterion is met. A failed assertion is a result to investigate, not proof by itself of a vulnerability.

**How to explain it in an interview:** ?I selected a method for the evaluation objective, set a measurable pass criterion, used controlled synthetic data, and made the check repeatable. This notebook demonstrates the approach; it does not claim that DeepTeam or a live model was executed.?


```python
import json

# Fictional, local-only observation.
observed = {
    "results": [
        "Demo status is ready.",
        " demo status is ready ",
        "DEMO STATUS IS READY."
    ]
}

normalized=["".join(ch for ch in x.casefold() if ch.isalnum()) for x in observed["results"]]
assert len(normalized)>=2 and len(set(normalized))==1
print("PASS 10/10: Scenario-set consistency")
```
