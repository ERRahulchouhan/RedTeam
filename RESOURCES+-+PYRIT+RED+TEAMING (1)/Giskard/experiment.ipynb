## Method 9: Robustness slice

**Method:** Robustness slice.

**Goal:** Normalize and compare benign response variations.

**Example:** Use a controlled, harmless fixture appropriate to this method; for example, robustness slice with a synthetic response or dataset.

**Expected result:** The next cell's assertion passes only when the stated criterion is met. A failed assertion is a result to investigate, not proof by itself of a vulnerability.

**How to explain it in an interview:** ?I selected a method for the evaluation objective, set a measurable pass criterion, used controlled synthetic data, and made the check repeatable. This notebook demonstrates the approach; it does not claim that Giskard or a live model was executed.?


```python
import json

# Fictional, local-only observation.
observed = {
    "results": [
        "Demo status is ready.",
        " demo status is ready ",
        "DEMO STATUS IS READY."
    ]
}

normalized=["".join(ch for ch in x.casefold() if ch.isalnum()) for x in observed["results"]]
assert len(normalized)>=2 and len(set(normalized))==1
print("PASS 9/10: Robustness slice")
```

## Method 10: Action policy slice

**Method:** Action policy slice.

**Goal:** Verify mock actions stay within an allowlist.

**Example:** Use a controlled, harmless fixture appropriate to this method; for example, action policy slice with a synthetic response or dataset.

**Expected result:** The next cell's assertion passes only when the stated criterion is met. A failed assertion is a result to investigate, not proof by itself of a vulnerability.

**How to explain it in an interview:** ?I selected a method for the evaluation objective, set a measurable pass criterion, used controlled synthetic data, and made the check repeatable. This notebook demonstrates the approach; it does not claim that Giskard or a live model was executed.?


```python
import json

# Fictional, local-only observation.
observed = {
    "allowed": [
        "lookup_demo_status"
    ],
    "executed": [
        "lookup_demo_status"
    ],
    "rejected": [
        "send_external_email"
    ]
}

assert set(observed["executed"])<=set(observed["allowed"])
assert set(observed["rejected"])-set(observed["executed"])
print("PASS 10/10: Action policy slice")
```
